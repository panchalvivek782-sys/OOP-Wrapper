```python
class Employee:
    def __init__(self, name, age, employee_id=None, salary=None):
        self.name = name
        self.age = age
        self._employee_id = employee_id
        self._salary = salary

    @property
    def employee_id(self):
        return self._employee_id

    @employee_id.setter
    def employee_id(self, value):
        self._employee_id = value

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        self._salary = value

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        if self._employee_id is not None:
            print(f"Employee ID: {self._employee_id}")
        if self._salary is not None:
            print(f"Salary: ${float(self._salary):.1f}")

    def __del__(self):
        pass


class Manager(Employee):
    def __init__(self, name, age, employee_id, salary, department):
        super().__init__(name, age, employee_id, salary)
        self.department = department

    def display(self):
        super().display()
        print(f"Department: {self.department}")


class Developer(Employee):
    def __init__(self, name, age, employee_id, salary, programming_language):
        super().__init__(name, age, employee_id, salary)
        self.programming_language = programming_language

    def display(self):
        super().display()
        print(f"Programming Language: {self.programming_language}")


def main():
    person = None
    employee = None
    manager = None

    while True:
        print("\n--- Python OOP Project: Employee Management System ---")
        print("1. Create a Person")
        print("2. Create an Employee")
        print("3. Create a Manager")
        print("4. Show Details")
        print("5. Exit")
        
        choice = input("\nEnter your choice: ")

        if choice == "1":
            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            person = Employee(name, age)
            print(f"\ Person created with name: {person.name} and age: {person.age}.")

        elif choice == "2":
            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            emp_id = input("Enter Employee ID: ")
            salary = float(input("Enter Salary: "))
            employee = Employee(name, age, emp_id, salary)
            print(f"\nEmployee created with name: {employee.name}, age: {employee.age}, ID: {employee.employee_id}, and salary: ${float(employee.salary):.1f}.")

        elif choice == "3":
            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            emp_id = input("Enter Employee ID: ")
            salary = float(input("Enter Salary: "))
            dept = input("Enter Department: ")
            manager = Manager(name, age, emp_id, salary, dept)
            print(f"\nManager created with name: {manager.name}, age: {manager.age}, ID: {manager.employee_id}, salary: ${float(manager.salary):.1f}, and department: {manager.department}.")

        elif choice == "4":
            print("\nChoose details to show:")
            print("1. Person")
            print("2. Employee")
            print("3. Manager")
            sub_choice = input("Enter your choice: ")

            if sub_choice == "1":
                if person:
                    print("\nPerson Details:")
                    person.display()
                else:
                    print("\nNo Person record found.")
            elif sub_choice == "2":
                if employee:
                    print("\nEmployee Details:")
                    employee.display()
                else:
                    print("\nNo Employee record found.")
            elif sub_choice == "3":
                if manager:
                    print("\nManager Details:")
                    manager.display()
                else:
                    print("\nNo Manager record found.")
            else:
                print("\nInvalid choice.")

        elif choice == "5":
            if issubclass(Manager, Employee) and issubclass(Developer, Employee):
                del person
                del employee
                del manager
                print("\nExiting the system. All resources have been freed.")
                print("Goodbye!")
                break
        else:
            print("\nInvalid option. Please try again.")

        print("\n--- Choose another operation ---")


if __name__ == "__main__":
    main()
```
